```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
# PR9 (Part A): K-Means Clustering - Elbow Method (Mall Customers Dataset)
file_path = 'Mall_Customers.csv'
df = pd.read_csv(file_path)
print("Dataset loaded successfully! Shape:", df.shape)
display(df.head(3))

# Selecting features: Annual Income (index 3) and Spending Score (index 4)
X = df.iloc[:, [3, 4]].values
```

```python
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

print("--- Inertia Values for k=1 to 10 ---")
for k, val in enumerate(inertia, start=1):
    print(f"k = {k}: Inertia = {val:.4f}")

plt.figure(figsize=(8, 4))
plt.plot(range(1, 11), inertia, marker='o', linestyle='-', color='purple')
plt.title('The Elbow Method (Optimal K Selection)')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```
